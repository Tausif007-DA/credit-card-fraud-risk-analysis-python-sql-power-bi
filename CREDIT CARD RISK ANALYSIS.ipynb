```python
from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = sin(dlat/2)**2 + cos(lat1)*cos(lat2)*sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))

    return R * c

df['distance'] = df.apply(lambda x: haversine(
    x['lat'], x['long'], x['merch_lat'], x['merch_long']), axis=1)
```

# Q1: What is the overall fraud distribution?

```python
# Set a stylish theme
sns.set_theme(style="whitegrid")

# Calculate counts and percentages
fraud_counts = df['is_fraud'].value_counts()
total = len(df)

# Initialize the figure
plt.figure(figsize=(10, 6))

# Create the barplot with a custom color palette
ax = sns.barplot(x=fraud_counts.index, y=fraud_counts.values, palette=['#4C72B0', '#C44E52'])

# Add detailing: Title and Labels
plt.title("Distribution of Fraudulent vs. Non-Fraudulent Transactions", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Transaction Type", fontsize=12)
plt.ylabel("Number of Transactions (Log Scale)", fontsize=12)
plt.xticks(ticks=[0, 1], labels=['Legit (0)', 'Fraud (1)'])

# --- THE MAGIC TOUCHES ---

# 1. Use Log Scale: Since fraud is usually < 1%, a log scale makes the fraud bar visible
ax.set_yscale("log")

# 2. Add Annotations (Labels on top of bars)
for p in ax.patches:
    height = p.get_height()
    percentage = (height / total) * 100
    ax.annotate(f'{int(height)}\n({percentage:.2f}%)', 
                (p.get_x() + p.get_width() / 2., height), 
                ha = 'center', va = 'center', 
                xytext = (0, 15), 
                textcoords = 'offset points',
                fontsize=11, fontweight='bold')

# Remove top and right spines for a cleaner look
sns.despine()

plt.tight_layout()
plt.show()
```

# Q5: Does Distance Between Customer & Merchant Indicate Fraud?

```python
# 1. DEBUG: Calculate the distance using the Haversine formula
def haversine(lat1, lon1, lat2, lon2):
    # Radius of the Earth in miles
    R = 3958.8 
    phi1, phi2 = np.radians(lat1), np.radians(lat2)
    dphi = np.radians(lat2 - lat1)
    dlambda = np.radians(lon2 - lon1)
    a = np.sin(dphi/2)**2 + np.cos(phi1) * np.cos(phi2) * np.sin(dlambda/2)**2
    return 2 * R * np.arctan2(np.sqrt(a), np.sqrt(1-a))

# Apply the calculation to create the missing 'distance' column
df['distance'] = haversine(df['lat'], df['long'], df['merch_lat'], df['merch_long'])

# 2. BEAUTIFY: Create bins and plot
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Create meaningful bins (e.g., 0-10 miles, 10-50, etc.) or use pd.cut
df['distance_bin'] = pd.cut(df['distance'], bins=5)
fraud_by_distance = df.groupby('distance_bin')['is_fraud'].mean()

# Plot using Seaborn for better detailing
ax = sns.barplot(x=fraud_by_distance.index.astype(str), y=fraud_by_distance.values, palette="magma")

# 3. ADD DETAILING
plt.title("Fraud Rate vs. Distance from Home", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Distance Range (Miles)", fontsize=12)
plt.ylabel("Average Fraud Rate", fontsize=12)
plt.xticks(rotation=30)

# Add percentage labels on top of bars
for p in ax.patches:
    ax.annotate(f'{p.get_height()*100:.2f}%', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 10), 
                textcoords='offset points', fontsize=10, fontweight='bold')

sns.despine()
plt.tight_layout()
plt.show()
```

# Q6: Which Age Group is More Prone to Fraud?

```python
from datetime import datetime

# 1. DEBUG: Convert dates and calculate Age
df['dob'] = pd.to_datetime(df['dob'])
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])

# Calculate age at the time of transaction
df['age'] = (df['trans_date_trans_time'] - df['dob']).dt.days // 365

# 2. Now create the age groups
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 25, 35, 50, 65, 100],
    labels=['0-25', '26-35', '36-50', '51-65', '65+']
)

# 3. Calculate fraud rate
fraud_by_age = df.groupby('age_group')['is_fraud'].mean()

# 4. BEAUTIFY: Plotting
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Use a professional color palette ('viridis' or 'coolwarm')
ax = sns.barplot(x=fraud_by_age.index, y=fraud_by_age.values, palette="viridis")

# Add detailing
plt.title("Fraud Vulnerability by Age Group", fontsize=16, fontweight='bold', pad=20)
plt.xlabel("Age Category", fontsize=12)
plt.ylabel("Fraud Rate (Percentage)", fontsize=12)

# 5. Add Percentage Labels on top
for p in ax.patches:
    ax.annotate(f'{p.get_height()*100:.2f}%', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                xytext=(0, 9), 
                textcoords='offset points',
                fontsize=11, fontweight='bold')

sns.despine()
plt.tight_layout()
plt.show()
```

# EXECUTIVE SUMMERY 

```python
import matplotlib.gridspec as gridspec

# --- 1. PREPARE DATA ---
# (Ensuring all derived columns exist)
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['hour'] = df['trans_date_trans_time'].dt.hour
df['dob'] = pd.to_datetime(df['dob'])
df['age'] = (df['trans_date_trans_time'] - df['dob']).dt.days // 365
df['age_group'] = pd.cut(df['age'], bins=[0, 25, 35, 50, 65, 100], labels=['0-25', '26-35', '36-50', '51-65', '65+'])

# --- 2. SET UP DASHBOARD LAYOUT ---
sns.set_theme(style="white")
fig = plt.figure(figsize=(20, 14))
gs = gridspec.GridSpec(3, 3, figure=fig)

# Global Title
fig.suptitle('EXECUTIVE SUMMARY: CREDIT CARD FRAUD RISK PROFILE', fontsize=26, fontweight='bold', y=0.95)

# --- 3. DONUT CHART: Overall Imbalance (Top Left) ---
ax1 = fig.add_subplot(gs[0, 0])
fraud_counts = df['is_fraud'].value_counts()
colors = ['#aec7e8', '#d62728']
ax1.pie(fraud_counts, labels=['Legit', 'Fraud'], autopct='%1.1f%%', startangle=90, 
        colors=colors, pctdistance=0.85, explode=[0, 0.2], textprops={'fontweight':'bold'})
center_circle = plt.Circle((0,0), 0.70, fc='white')
ax1.add_artist(center_circle)
ax1.set_title("Transaction Landscape", fontsize=15, fontweight='bold')

# --- 4. HORIZONTAL BAR: Top Risk Categories (Top Center/Right) ---
ax2 = fig.add_subplot(gs[0, 1:])
top_cats = df.groupby('category')['is_fraud'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=top_cats.values * 100, y=top_cats.index, ax=ax2, palette="Reds_r")
ax2.set_title("Top 10 High-Risk Categories (by %)", fontsize=15, fontweight='bold')
ax2.set_xlabel("Fraud Probability (%)")
for i, v in enumerate(top_cats.values):
    ax2.text(v*100 + 0.05, i, f'{v*100:.2f}%', va='center', fontweight='bold', color='darkred')

# --- 5. AREA CHART: Temporal Trends (Middle Row - Spans Width) ---
ax3 = fig.add_subplot(gs[1, :])
fraud_by_hour = df.groupby('hour')['is_fraud'].mean()
ax3.plot(fraud_by_hour.index, fraud_by_hour.values, color='#d62728', linewidth=4, marker='o', markersize=10)
ax3.fill_between(fraud_by_hour.index, fraud_by_hour.values, color='#d62728', alpha=0.15)
ax3.set_title("Hourly Fraud Risk: When do Fraudsters Strike?", fontsize=15, fontweight='bold')
ax3.set_xticks(range(0, 24))
ax3.set_ylabel("Risk Factor")
# Annotation for peak
peak_h = fraud_by_hour.idxmax()
ax3.annotate('CRITICAL PEAK', xy=(peak_h, fraud_by_hour.max()), xytext=(peak_h-3, fraud_by_hour.max()*1.1),
             arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, fontweight='bold', color='red')

# --- 6. BAR CHART: Demographic Risk (Bottom Left) ---
ax4 = fig.add_subplot(gs[2, 0])
age_risk = df.groupby('age_group')['is_fraud'].mean()
sns.barplot(x=age_risk.index, y=age_risk.values * 100, ax=ax4, palette="viridis")
ax4.set_title("Risk by Age Group", fontsize=15, fontweight='bold')
ax4.set_ylabel("Fraud Rate (%)")

# --- 7. VIOLIN PLOT: Value Distribution (Bottom Center/Right) ---
ax5 = fig.add_subplot(gs[2, 1:])
# Clipping amount for visibility in the summary
sns.violinplot(x='is_fraud', y='amt', data=df[df['amt'] < 1000], ax=ax5, palette=['#aec7e8', '#d62728'], split=True)
ax5.set_title("Transaction Value Analysis (Transactions < $1000)", fontsize=15, fontweight='bold')
ax5.set_xticklabels(['Legitimate', 'Fraudulent'])
ax5.set_ylabel("Amount ($)")


# --- 8. FINAL POLISH ---
plt.tight_layout(rect=[0, 0.03, 1, 0.92])
sns.despine()
plt.show()
```
